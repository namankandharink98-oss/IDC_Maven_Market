```sql
USE CATALOG maven_catalog;
```

```sql
-- Data Engineers: full control on schemas
GRANT USAGE ON CATALOG maven_catalog TO grp_data_engineers;
GRANT CREATE ON CATALOG maven_catalog TO grp_data_engineers;
GRANT ALL PRIVILEGES ON SCHEMA maven_catalog.bronze_schema TO grp_data_engineers;
GRANT ALL PRIVILEGES ON SCHEMA maven_catalog.silver_schema TO grp_data_engineers;
GRANT ALL PRIVILEGES ON SCHEMA maven_catalog.gold_schema TO grp_data_engineers;
```

```sql
-- Analysts: read-only on gold
GRANT USAGE ON CATALOG maven_catalog TO grp_analysts;
GRANT USE SCHEMA ON SCHEMA maven_catalog.gold_schema TO grp_analysts;
GRANT SELECT ON SCHEMA maven_catalog.gold_schema TO grp_analysts;
--Allows analysts to read tables and views in the gold schema only.They cannot create, modify, or delete objects—purely read-only.
```

```sql
-- Regional managers: only select on secured views (RLS)
GRANT USAGE ON CATALOG maven_catalog TO regional_manager;
GRANT USE SCHEMA ON SCHEMA maven_catalog.gold_schema TO regional_manager;
GRANT SELECT ON SCHEMA maven_catalog.gold_schema TO grp_analysts;
GRANT SELECT ON VIEW maven_catalog.gold_schema.secured_fact_sales TO regional_manager;
GRANT SELECT ON VIEW maven_catalog.gold_schema.secured_fact_returns TO regional_manager;
```

```sql
--Executives: only aggregate views
GRANT USAGE ON CATALOG maven_catalog TO grp_executives;
GRANT SELECT ON VIEW maven_catalog.gold_schema.agg_customer_lifetime TO grp_executives;
GRANT SELECT ON VIEW maven_catalog.gold_schema.agg_daily_sales TO grp_executives;
GRANT SELECT ON VIEW maven_catalog.gold_schema.agg_monthly_sales TO grp_executives;
GRANT SELECT ON VIEW maven_catalog.gold_schema.agg_product_performance TO grp_executives;
GRANT SELECT ON VIEW maven_catalog.gold_schema.agg_store_performance TO grp_executives;
```
