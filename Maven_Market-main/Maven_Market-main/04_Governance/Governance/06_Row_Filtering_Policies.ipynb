```sql
use catalog maven_catalog;
```

```sql
use schema governance
```

```sql
CREATE OR REPLACE FUNCTION maven_catalog.governance.rls_sales_region_filter(sales_region_col STRING)
RETURNS BOOLEAN
RETURN
(
    /* 1. Bypass UC groups: admins, data engineers */
    is_account_group_member('admins')
    OR is_account_group_member('data_engineers')
    OR
    /* 2. Regional Manager filter */
    sales_region_col IN (
        SELECT allowed_sales_region
        FROM maven_catalog.governance.region_manager_access
        WHERE user_email = current_user()
    )
);
```
