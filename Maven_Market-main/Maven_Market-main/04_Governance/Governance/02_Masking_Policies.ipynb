```sql
USE CATALOG maven_catalog;
USE SCHEMA gold_schema;
```

```sql
CREATE OR REPLACE FUNCTION pii_mask_policy(val STRING)
RETURN CASE
  -- Bypass masking for Admins or specific power users
  WHEN is_account_group_member('admin') THEN val
  -- Group based masking
  WHEN is_account_group_member('grp_data_engineers') THEN val
  WHEN is_account_group_member('grp_analyst') THEN NULL
  WHEN is_account_group_member('regional_manager') THEN NULL
  ELSE NULL
END;
```

```sql
CREATE OR REPLACE FUNCTION custm_acc_partial_mask(pc STRING)
RETURN CASE
  WHEN is_account_group_member('admin') THEN pc
  WHEN is_account_group_member('grp_data_engineers') THEN pc
  WHEN is_account_group_member('grp_analysts') OR is_account_group_member('regional_manager') 
       THEN CONCAT(SUBSTR(pc, 1, 3), '****')
  ELSE NULL
END;
```

```sql
CREATE OR REPLACE FUNCTION numeric_mask_policy(val DOUBLE)
RETURN CASE
  WHEN is_account_group_member('admin') THEN val
  WHEN is_account_group_member('grp_data_engineers') THEN val
  ELSE NULL -- Analysts and Managers see NULL for numeric privacy
END;
```
