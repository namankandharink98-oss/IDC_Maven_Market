```python
Store Performance Leaderboard
```

```python
%sql
SELECT 
  store_name,
  store_city,
  ROUND(SUM(total_revenue),2) AS total_revenue,
  SUM(total_transactions) AS total_transactions,
  ROUND(SUM(total_profit) / SUM(total_revenue) * 100, 2) AS profit_margin_pct
FROM maven_catalog.gold_schema.agg_store_performance
GROUP BY store_name, store_city
ORDER BY total_revenue DESC;
```

Regional Sales Trend (YoY)

```python
%sql
SELECT 
  year_month,
  ROUND(SUM(total_revenue),2) AS current_revenue,
  ROUND(LAG(SUM(total_revenue), 12) OVER (ORDER BY year_month),2) AS last_year_revenue
FROM maven_catalog.gold_schema.agg_monthly_sales
GROUP BY year_month
ORDER BY year_month DESC;
```

Store Efficiency (Revenue per Square Foot)

```python
%sql
SELECT 
  store_name,
  total_sqft,
  ROUND(total_revenue,2),
  ROUND(total_profit,2),
  revenue_per_sqft
FROM maven_catalog.gold_schema.agg_store_performance
ORDER BY revenue_per_sqft DESC;
```

Regional Product Mix

```python
%sql
SELECT 
  p.product_brand,
  ROUND(SUM(f.total_revenue),2) AS total_revenue
FROM maven_catalog.gold_schema.fact_sales f
JOIN maven_catalog.gold_schema.dim_products p ON f.product_id = p.product_key
GROUP BY p.product_brand
ORDER BY total_revenue DESC;
```
