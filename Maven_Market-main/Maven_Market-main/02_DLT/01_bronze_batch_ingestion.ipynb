###Transactions data

```python
import dlt
from pyspark.sql.functions import (
    input_file_name, current_timestamp, lit
)

@dlt.table(
    name="maven_catalog.bronze_schema.brz_transactions",
    comment="Raw transactions from ADLS CSV (1997 + 1998)",
    table_properties={"quality": "bronze"}
)
def brz_transactions():
    return (
        spark.readStream
        .format("cloudFiles")
        .option("cloudFiles.format", "csv")
        .option("cloudFiles.inferColumnTypes", "true")
        .option("header", "true")
        .option("cloudFiles.schemaHints",
                "transaction_date STRING, stock_date STRING, "
                "product_id INT, customer_id INT, "
                "store_id INT, quantity INT")
        .load("abfss://mavenmarket@sgmavenmarkets.dfs.core.windows.net/raw/transactions/")
        .withColumn("_ingestion_timestamp", current_timestamp())
        .withColumn("_source_system", lit("csv_batch"))
    )
```
